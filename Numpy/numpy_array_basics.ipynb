```python
"""
- numpy arrays are similar to C/C++ arrays i.e they are of a fixed data type unlike Python
- less flexible but more efficient for storing & manipulating data unlike Python's dynamic/flexible lists
"""

import numpy as np

##-------------- WAYS TO CREATE NUMPY ARRAYS-----------------
# _____________________________________________________________________________________________
#1. Creating arrays from python lists
np.array([1,4,2,5,3])
```

```python
#if types of elements in the array don't match, Numpy 'upcast's them (int to float in this case:)
np.array([3.142, 4, 2,3])
```

```python
#can also explicitly state data type using dtype keyword:
np.array([1,2,3,4], dtype = np.float32)
```

```python
#multi-dimensional arrays can also exist in numpy unlike 1D python lists:
np.array([range(i, i+3) for i in [2,4,6]])
```

```python
# ________________________________________________________________________________________________________________
# 2. Creating arrays from scratch
np.zeros(3, dtype = int)
```

```python
#Array attributes: ndim, size, shape, dtype
rng = np.random.default_rng(seed = 42)
x1 = rng.integers(10, size = 6)   #dimension 1, array having integers 0 to 9
x2 = rng.integers (10, size = (3,3))  #2x2 dim
x3 = rng.integers (10, size = (2,3,5)) #3x3 dim
```

```python
print(f"x1: dim {x1.ndim}, size {x1.size}, shape {x1.shape}, dtype {x1.dtype}")
print(f"x2: dim {x2.ndim}, size {x2.size}, shape {x2.shape}, dtype {x2.dtype}")
print(f"x3: dim {x3.ndim}, size {x3.size}, shape {x3.shape}, dtype {x3.dtype}")
```

```python
"""
- One common use of slicing is that along with indexing, a single row or column can be accessed
- Important to note that these sliced arrays act as VIEWS rather than the copies of array. Modification in sliced arrays results 
in modification in og base array as well.
- .copy() method for creating copies if needed for use
"""
x2[0, :2]   #first row (index = 0) & columns sliced till 2nd column
```

```python
x2
```

```python
x2_sub = x2[0:2]
x2_sub[0] = 8
x2
```

```python
x2_sub_copy = x2[0:2].copy()
x2_sub_copy[0] = 12     #modification in copy has no effect on the og array
x2
```

```python
#_______________________________________________________________________________________________________________________
# Reshaping Arrays -> reshape() changes the arrangement/shape attribute of data in an array.. ONLY possible if the og size and the reshape size match
np.arange(1,10)    
```

```python

#_______________________________________________________________________________________________________________________

# Array concatenation (concatenate(), vstack(), hstack()) & splitting (split(), vsplit(), hsplit())
x = [1,2,3]
y = [4,5,6]
z = [7,8,9]
array = np.concatenate([x,y,z])
array
```

```python
grid = np.eye(3, dtype = int)
grid
```

```python
np.concatenate([grid, grid])     #concatenating 2d arrays
```

```python
xy_vert = np.vstack([x,y])
xy_vert
```

```python
xy_hor = np.hstack([x,y])
xy_hor
```

```python
#For splitting, we can pass list of indices as splitting point.. NOTE: N splitting points result in N+1 arrays
x = np.arange(1,10)
x
```

```python
x1, x2, x3 = np.split(x, [3,5])
print(x1,x2,x3)
```

```python
y = np.arange(0,16).reshape(4,4)
y
```

```python
upper, lower = np.vsplit(y, [2])   #vertical split -> split into upper lower 
print(upper)
print(lower)
```

```python
left, right = np.hsplit(y,[2])     #horizontal split-> split into left right
print(left)
print(right)
```
